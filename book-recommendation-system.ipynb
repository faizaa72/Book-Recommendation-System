```python
ratings_with_name = ratings.merge(books,on='ISBN')
ratings_with_name
```

## Popularity Based System

```python
num_rating_df=ratings_with_name.groupby('Book-Title')['Book-Rating'].count().reset_index()
num_rating_df.rename(columns=({'Book-Rating':'num_rating'}),inplace=True)
num_rating_df
```

```python
avg_rating_df=ratings_with_name.groupby('Book-Title')['Book-Rating'].mean(numeric_only=True).reset_index()
avg_rating_df.rename(columns=({'Book-Rating':'avg_rating'}),inplace=True)
avg_rating_df
```

```python
popular_df= num_rating_df.merge(avg_rating_df,on='Book-Title')
popular_df
```

```python
popular_df=popular_df[popular_df['num_rating']>250].sort_values('avg_rating',ascending=False).head(50)
popular_df
```

```python
popular_df=popular_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title','Book-Author','Year-Of-Publication','num_rating','avg_rating','Image-URL-M']]
```

 ## Collaborative Based System

```python
x=ratings_with_name.groupby('User-ID').count()['Book-Rating'] > 200
high_rating_index = x[x].index
high_rating_index
```

```python
filtered_df=ratings_with_name[ratings_with_name['User-ID'].isin(high_rating_index)]
```

```python
y=filtered_df.groupby('Book-Title').count()['Book-Rating']>=50
book_high_rating=y[y].index
```

```python
final_result=filtered_df[filtered_df['Book-Title'].isin(book_high_rating)]
```

```python
pt=final_result.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores=cosine_similarity(pt)
```

```python
similarity_scores.shape
```

```python
def recommend(book_name):
    index=np.where(pt.index==book_name)[0][0]
    similar_items=sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[1:6]

    for i in similar_items:
        print(pt.index[i[0]])
```

```python
recommend('A Prayer for Owen Meany')
```
